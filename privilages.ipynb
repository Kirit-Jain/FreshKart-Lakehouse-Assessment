```python
%sql
-- For fk_analysts (Gold SELECT only)
GRANT USE CATALOG ON CATALOG identifier(:catalog) TO `fk_analysts`;
GRANT USE SCHEMA ON SCHEMA identifier(:catalog).gold TO `fk_analysts`;
GRANT SELECT ON SCHEMA identifier(:catalog).gold TO `fk_analysts`;

-- For fk_engineers (broader privileges)
GRANT USE CATALOG ON CATALOG identifier(:catalog) TO `fk_engineers`;
GRANT USE SCHEMA ON SCHEMA identifier(:catalog).bronze TO `fk_engineers`;
GRANT USE SCHEMA ON SCHEMA identifier(:catalog).silver TO `fk_engineers`;
GRANT USE SCHEMA ON SCHEMA identifier(:catalog).gold TO `fk_engineers`;
GRANT USE SCHEMA ON SCHEMA identifier(:catalog).ops TO `fk_engineers`;

GRANT SELECT, MODIFY ON SCHEMA identifier(:catalog).bronze TO `fk_engineers`;
GRANT SELECT, MODIFY ON SCHEMA identifier(:catalog).silver TO `fk_engineers`;
GRANT SELECT, MODIFY ON SCHEMA identifier(:catalog).gold TO `fk_engineers`;
GRANT SELECT, MODIFY ON SCHEMA identifier(:catalog).ops TO `fk_engineers`;

GRANT READ VOLUME ON VOLUME identifier(:catalog).landing.raw TO `fk_engineers`;
GRANT WRITE VOLUME ON VOLUME identifier(:catalog).landing.raw TO `fk_engineers`;
```
